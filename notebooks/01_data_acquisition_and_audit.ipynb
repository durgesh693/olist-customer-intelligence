# Stage 3: Data Requirements & Data Acquisition Audit
**Objective:** Verify that all required Olist Brazilian E-Commerce raw CSV files are present, uncorrupted, and accurately ingested with their exact row counts, column counts, and disk sizes.

```python
import os
import pandas as pd
from pathlib import Path

# Relative path from 'notebooks/' directory to 'data/raw/'
RAW_DATA_PATH = Path("../data/raw")

# List of expected files as per Master Architecture
EXPECTED_FILES = [
    "olist_customers_dataset.csv",
    "olist_orders_dataset.csv",
    "olist_order_items_dataset.csv",
    "olist_order_payments_dataset.csv",
    "olist_order_reviews_dataset.csv",
    "olist_products_dataset.csv",
    "olist_sellers_dataset.csv",
    "olist_geolocation_dataset.csv",
    "product_category_name_translation.csv"
]

print(f"Checking files in: {RAW_DATA_PATH.resolve()}")
```

```python
audit_records = []

for file_name in EXPECTED_FILES:
    file_path = RAW_DATA_PATH / file_name
    
    if file_path.exists():
        # Fast load to capture exact shapes
        df_temp = pd.read_csv(file_path)
        file_size_mb = os.path.getsize(file_path) / (1024 * 1024)
        
        audit_records.append({
            "File Name": file_name,
            "Rows": int(df_temp.shape[0]),
            "Columns": int(df_temp.shape[1]),
            "File Size (MB)": round(file_size_mb, 2),
            "Status": "Available"
        })
    else:
        audit_records.append({
            "File Name": file_name,
            "Rows": None,
            "Columns": None,
            "File Size (MB)": 0.0,
            "Status": "Missing"
        })

# Render structured summary
audit_summary = pd.DataFrame(audit_records)
display(audit_summary)
```
